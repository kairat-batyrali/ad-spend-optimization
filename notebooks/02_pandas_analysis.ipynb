```python
import pandas as pd

daily = pd.read_csv("../data/daily_performance.csv", parse_dates=["date"])
campaigns = pd.read_csv("../data/campaigns.csv")

daily.shape
```

```python
# Question: same duplicates as in SQL?
# Grain: one row per campaign per day
# Check: exact-copy rows removed; expect 13048 rows
daily_clean = daily.drop_duplicates()
len(daily_clean)
```

```python
# Question: which campaigns return the most revenue per dollar?
# Grain: one row per campaign (daily rows summed first)
# Check: ROAS and CPA are ratios of sums; merge is validated one_to_one
by_campaign = (
    daily_clean.groupby("campaign_id", as_index=False)
    .agg(spend=("spend", "sum"),
         revenue=("revenue", "sum"),
         conversions=("conversions", "sum"))
)
by_campaign["ROAS"] = by_campaign["revenue"] / by_campaign["spend"]
by_campaign["CPA"] = by_campaign["spend"] / by_campaign["conversions"]

result = by_campaign.merge(
    campaigns[["campaign_id", "campaign_name", "channel"]],
    on="campaign_id",
    how="left",
    validate="one_to_one",
)

result.sort_values("roas", ascending=False).head(20)
```
